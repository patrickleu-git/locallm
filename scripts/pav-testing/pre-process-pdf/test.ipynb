# Testing Document Loaders

```python
import pprint
```

Paths and prompts

```python
DATA_PATH = "C:/Users/patri/Documents/Learning Sources/github/locallm/data/pdfs/pav"

IMAGE_PROMPT = """
Please inspect the image carfully. First, extract the text on the picture. Describe what you see.
Second, try to put what you see into larger context, i.e., try to interpret the graphs or maps on the image.
"""
```

### PyPDFLoader

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.document_loaders.parsers import LLMImageBlobParser
from langchain_ollama import ChatOllama
```

```python
loader_pypdf = PyPDFLoader(
    file_path=DATA_PATH + "/extract.pdf",
    mode = "page",
    extract_images=True,
    images_parser=LLMImageBlobParser(
        model=ChatOllama(
            model="qwen2.5vl:7b",
            temperature=1,
            num_predict=1024
        ),
        prompt=IMAGE_PROMPT
    ),
    images_inner_format="markdown-img",
    extraction_mode="layout"
)

docs_pypdf = loader_pypdf.load()

print(f"Number of docs: {len(docs_pypdf)}")
pprint.pp(docs_pypdf[0].metadata)
pprint.pp(docs_pypdf[0].page_content)
```

Performance: Quite fast, but the loader cannot differ between an actual linebreaks and sentences that continue on the next line... 

Moreover, the coloured textboxes are not directly interpreted as text, and again, new and continuing lines are not spotted (but may identifiable by the "unordered-list-dot"). 

Somehow images are not captured at all??

### PyMuPDF

```python
from langchain_community.document_loaders import PyMuPDFLoader
```

```python
loader_pymu = PyMuPDFLoader(
    file_path=DATA_PATH + "/extract.pdf",
    mode = "page",
    extract_tables="markdown",
    extract_images=True,
    images_parser=LLMImageBlobParser(
        model=ChatOllama(
            model="gemma3:4b",
            temperature=1,
            num_predict=1024
        ),
        prompt=IMAGE_PROMPT
    ),
    images_inner_format="markdown-img"
)

docs_pymu = loader_pymu.load()

print(f"Number of docs: {len(docs_pymu)}")
pprint.pp(docs_pymu[0].metadata)
pprint.pp(docs_pymu[0].page_content)
```

Performance: Significantly slower than PyPDFLoader; the loader cannot differ between an actual linebreaks and sentences that continue on the next line... 

Moreover, the coloured textboxes are interpreted as tables, and again, new and continuing lines are not spotted (but may identifiable by the "unordered-list-dot"). 

Somehow images are not captured at all??
