Das Agglomerationsprogramm Winterthur und Umgebung, 5. Generation, berücksichtigt **Intermodalität** und **kombinierte Mobilität** – insbesondere in Bezug auf die Verknüpfung von Verkehrsmitteln an geeigneten Orten –, jedoch **nicht in einer zentralen, umfassenden Weise**. Die Betrachtung der Massnahmen im Kontext des Programms ergibt folgende Einschätzung:
### 1. **Optimierung des regionalen und städtischen Busnetzes (Mn-Nr. ÖV9)**
- **Intermodalität**: Die Massnahme zielt auf die **Optimierung des Busnetzes** ab, was indirekt zur Verbesserung der Verknüpfung mit anderen Verkehrsmitteln beitragen kann. Allerdings wird **keine explizite Berücksichtigung von Intermodalität** oder kombinierter Mobilität (z. B. MIV/ÖV/LV/Sharing-Angebote) genannt.
- **Verknüpfung an geeigneten Orten**: Es wird nicht ausdrücklich erwähnt, ob **Verknüpfungspunkte** oder **kombinierte Angebote** an strategischen Orten (z. B. Bahnhöfe, Züge, etc.) berücksichtigt werden.
- **Kontext**: Die Massnahme ist als **Daueraufgabe** aus der AP3 fortgeführt worden, was darauf hindeutet, dass die bisherigen Schritte in dieser Richtung nicht ausreichend waren, um eine umfassende Intermodalität zu erreichen.

### 2. **Zentrumserschliessung – Neuhegi-Grüze inkl. FlaMa (Mn-Nr. MIV1)**
- **Intermodalität**: Diese Massnahme ist spezifisch auf die **Erschliessung des Zentrums** und die **Flächenentwicklung** ausgerichtet. Sie ist **nicht direkt mit Intermodalität** oder kombinierter Mobilität verknüpft.
- **Verknüpfung an geeigneten Orten**: Die Massnahme bezieht sich auf eine **Weiterentwicklung der Massnahme GV10 aus der 3. Generation**, die ursprünglich aufgrund mangelhaften Reifegrades und Kosten-Nutzen-Verhältnisses zurückgestuft wurde. Es wird **keine explizite Berücksichtigung von Intermodalität** oder kombinierter Mobilität genannt.
- **Kontext**: Die Massnahme ist **nicht zentral auf Intermodalität ausgerichtet**, sondern eher auf die **physische Erschliessung** und **Flächenplanung**.  

### 3. **Zusammenfassung: Intermodalität im AP5**
- **Berücksichtigung**: Die Massnahmen des AP5 **berücksichtigen Intermodalität indirekt**, z. B. durch die **Optimierung von Verkehrsnetzen**, aber **keine konkreten Massnahmen zur Verbesserung der Verknüpfung von Verkehrsmitteln an strategischen Orten** sind explizit vorgesehen.
- **Fehlende Fokussierung**: Es fehlt eine **strategische Ausrichtung auf kombinierte Mobilität** (MIV/ÖV/LV/Sharing) und eine **konkrete Planung zur Verbesserung der Intermodalität**. Die Massnahmen sind **nicht in der Weise ausgearbeitet**, dass sie eine **systematische Verbesserung der Intermodalität** im Agglomerationsraum garantieren.

### Fazit:
Das **Agglomerationsprogramm Winterthur und Umgebung, 5. Generation, berücksichtigt Intermodalität und kombinierte Mobilität nur indirekt** und **nicht in einer zentralen, umfassenden Weise**. Es fehlen **konkrete Massnahmen zur Verknüpfung von Verkehrsmitteln an strategischen Orten**, insbesondere für MIV/ÖV/LV/Sharing-Angebote. Die bisherigen Massnahmen sind **nicht ausreichend auf die Förderung von Intermodalität ausgerichtet**, und dies bleibt ein **Kritikpunkt** im Kontext der strategischen Evaluation.
